```
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("Titanic-Spark-ML").master("local[*]").getOrCreate()         #Yapıyı kurmak için kullandık.
print("Spark başarıyla başlatıldı.")
```

```
dataframe=spark.read.csv("titanic.csv",header=True,inferSchema=True) #veriyi sparkla çekmek için
```

```
dataframe=dataframe.select("Survived","Pclass","Sex","Age","Fare")
```

```
dataframe=dataframe.dropna()
```

```
from pyspark.ml.feature import StringIndexer, OneHotEncoder
from pyspark.ml.feature import VectorAssembler
from pyspark.ml.classification import LogisticRegression
```

```
sex_indexer=StringIndexer(inputCol="Sex",outputCol="SexIndex")
dataframe=sex_indexer.fit(dataframe).transform(dataframe)
```

```
if "features" in dataframe.columns:
    dataframe = dataframe.drop("features")
    print("The 'features' column was successfully removed.")
else:
    print("The 'features' column does not exist in the DataFrame.")
```

```
dataframe.show(5)
```

```
assembler=VectorAssembler(inputCols=["Survived","Pclass","SexIndex","Age","Fare"],outputCol="features")
```

```
dataframe=assembler.transform(dataframe)
dataframe.select("features","Survived").show(5)
```

```
if "features" in dataframe.columns:
    print("The 'features' column already exists in the DataFrame.")
else:
    print("The 'features' column does not exist in the DataFrame.")
```

```
train_data,test_data=dataframe.randomSplit([0.8,0.2],seed=42)
```

```
from pyspark.ml.classification import LogisticRegression
```

```
lr=LogisticRegression(labelCol="Survived",featuresCol="features")
```

```
model=lr.fit(train_data)
```

```

predictions=model.transform(test_data)
predictions.select("Survived","prediction","probability").show(301)
```
